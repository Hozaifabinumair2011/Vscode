```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv('Titanic Dataset (3).csv')
data.head(5)
```

```python
data.isnull().sum()
```

```python
#quartile
age_q1 = np.quantile(data["Age"],0.25)
age_q2 = np.quantile(data["Age"],0.50)
age_q3 = np.quantile(data["Age"],0.75)

print("Age Quartiles - ")
print("Q1 -",age_q1)
print("Q2 -",age_q2)
print("Q3 -",age_q3)
```

```python
#Interquartile Range
IQR_age = age_q3 - age_q1
print("Interquartile Range :", IQR_age)
```

```python
plt.hist(data["Age"])
plt.ylabel("Count of Passengers")
plt.xlabel("Age")
```

```python
#Quartile2
fare_q1 = np.quantile(data["Fare"],0.25)
fare_q2 = np.quantile(data["Fare"],0.50)
fare_q3 = np.quantile(data["Fare"],0.75)

print("Fare Quartiles -")
print("Q1 -",fare_q1)
print("Q2 -",fare_q2)
print("Q3 -",fare_q3)
```

```python
#Interquartile Range2
IQR_fare = fare_q3 - fare_q1
print("Interquartile Range :", IQR_fare)
```

```python
bins = np.arange(0,250,20)
plt.hist(data["Fare"], bins=np.arange(1,250,20))
plt.ylabel("Count of Passengers")
plt.xlabel("Fare")
plt.xticks(bins)
```

```python
plt.boxplot(data["Age"])
plt.title("Age distribution")
plt.show()
```

```python
plt.boxplot(data["Pclass"])
plt.title("Passenger class distribution")
plt.show()
```
